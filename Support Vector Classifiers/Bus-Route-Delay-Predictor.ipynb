```python
import pandas as pd
from matplotlib import pyplot as plt
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import accuracy_score, recall_score, precision_score, classification_report, confusion_matrix, f1_score
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

route_df = pd.read_csv("Bus-Route-Delay-Prediction.csv")
route_df = route_df.dropna()
route_df['DelayClass'] = route_df['DelayClass'].map({"OnTime":1,"Delayed":2,"SeverelyDelayed":3})
```

```python
X = route_df.drop(columns="DelayClass")
Y = route_df["DelayClass"]

x_train, x_test, y_train, y_test= train_test_split(X,Y, test_size=0.2, random_state=65)
```

```python
Model = make_pipeline(
    StandardScaler(),
    SVC(kernel="rbf", probability=True)
)
params = {"C":[2.4, 1.3, 1.7, 1.8], "random_state":[42,56,78,32]}
grid = GridSearchCV(Model, params, cv=9)
grid.train(x_train, y_train)
```

```python
y_pred = grid.predict([[12,38.5,8,10]])
print(f"Accuracy Score = {accuracy_score(y_pred, y_test)}")
print(f"Classification Report = {classification_report(y_pred, y_test)}")
print(f"Confusion Matrix = {confusion_matrix(y_pred, y_test)}")
print(f"Precision Score = {precision_score(y_pred, y_test)}")
print(f"Recall Score = {recall_score(y_pred, y_test)}")
print(f"F1 Score = {f1_score(y_pred, y_test)}")
print()
try:
    if y_pred[0] == 1:
        print(f"Bus Route Delay Status = On Time")
    elif y_pred[0] == 1:
        print(f"Bus Route Delay Status = Delayed")
    elif y_pred[0] == 1:
        print(f"Bus Route Delay Status = Severely Delayed")
except BaseException as e:
    print(f"Error = {e}")                    
```

```python
colors = {
    "OnTime":"red",
    "Delayed":"purple",
    "SeverelyDelayed":"orange"
}
point_colors = y_pred.map(colors)
plt.figure(figsize=(17,6))
plt.scatter(x_test[X.columns], c=colors)
plt.title("Actual Bus Route Delay Statuses vs Predicted Bus Route Delay Statuses")
plt.xlabel("Actual Bus Route Delay Statuses")
plt.ylabel("Predicted Bus Route Delay Statuses")
plt.show()
```
