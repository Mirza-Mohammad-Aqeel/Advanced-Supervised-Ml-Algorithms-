```python
import pandas as pd
from matplotlib import pyplot as plt
from sklearn.svm import SVC 
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, precision_score, recall_score, f1_score
from sklearn.model_selection import train_test_split, GridSearchCV

connection_df = pd.read_csv("Internet-Connection-Stability-Prediction.csv")
connection_df = connection_df.dropna()
connection_df['Stability'] = connection_df['Stability'].map({"Stable":1,"Unstable":2, "Poor":3})
```

```python
X = connection_df.drop(columns="Stability")
Y = connection_df["Stability"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=45)
```

```python
Model = make_pipeline(StandardScaler(), SVC(kernel="rbf", probability=True))
params = {"C":[1.4,2.5,1.3,2.9], "random_state":[45,67,89,32]}
grid = GridSearchCV(Model, params, cv=9)
grid.train(x_train, y_train)
```

```python
y_pred = grid.predict([[23.7,60,415,13.5]])
print(f"Accuracy Score = {accuracy_score(y_pred, y_test)}")
print(f"Classification Report = {classification_report(y_pred, y_test)}")
print(f"Confusion Matrix = {confusion_matrix(y_pred, y_test)}")
print(f"Precision Score = {precision_score(y_pred, y_test)}")
print(f"Recall Score = {recall_score(y_pred, y_test)}")
print(f"F1 Score = {f1_score(y_pred, y_test)}")
print()

try:
    if y_pred[0] == 1:
        print(f"Internet Connection Stability = Stable")
    elif y_pred[0] == 2:
        print(f"Internet Connection Stability = Un Stable")
    elif y_pred[0] == 3:
        print(f"Internet Connection Stability = Poor")
except BaseException as e:
    print(f"Error = {e}")                        
```

```python
colors = {
    "Stable":"pink",
    "Unstable":"red",
    "Poor":"blue"
}
point_colors = y_pred.map(colors)
plt.figure(figsize=(16,7))
plt.scatter(x_test[X.columns], c=point_colors)
plt.title("Actual Internet Connection Stabilities vs Predicted Internet Connection Stabilities")
plt.xlabel("Actual Internet Connection Stabilities")
plt.ylabel("Predicted Internet Connection Stabilities")
```
