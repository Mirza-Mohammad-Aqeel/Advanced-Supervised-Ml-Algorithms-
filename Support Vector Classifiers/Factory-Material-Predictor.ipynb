```python
import pandas as pd
from matplotlib import pyplot as plt
from sklearn.svm import SVC 
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, precision_score, recall_score, f1_score
from sklearn.model_selection import train_test_split, GridSearchCV

factory_material_df = pd.read_csv("Factory-Material-Prediction.csv")
factory_material_df = factory_material_df.dropna()
factory_material_df['MaterialType'] = factory_material_df['MaterialType'].map({"TypeA":1,"TypeB":2, "TypeC":3})
```

```python
X = factory_material_df.drop(columns="MaterialType")
Y = factory_material_df["MaterialType"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=45)
```

```python
Model = make_pipeline(StandardScaler(), SVC(kernel="rbf", probability=True))
params = {"C":[1.4,2.5,1.3,2.9], "random_state":[45,67,89,32]}
grid = GridSearchCV(Model, params, cv=9)
grid.train(x_train, y_train)
```

```python
y_pred = grid.predict([[23.7,60,415,13.5]])
print(f"Accuracy Score = {accuracy_score(y_pred, y_test)}")
print(f"Classification Report = {classification_report(y_pred, y_test)}")
print(f"Confusion Matrix = {confusion_matrix(y_pred, y_test)}")
print(f"Precision Score = {precision_score(y_pred, y_test)}")
print(f"Recall Score = {recall_score(y_pred, y_test)}")
print(f"F1 Score = {f1_score(y_pred, y_test)}")
print()

try:
    if y_pred[0] == 1:
        print(f"Factory Material Type = Type A")
    elif y_pred[0] == 2:
        print(f"Factory Material Type = Type B")
    elif y_pred[0] == 3:
        print(f"Factory Material Type = Type C")
except BaseException as e:
    print(f"Error = {e}")                        
```

```python
colors = {
    "TypeA":"orange",
    "TypeB":"red",
    "TypeC":"blue"
}
point_colors = y_pred.map(colors)
plt.figure(figsize=(16,7))
plt.scatter(x_test[X.columns], c=point_colors)
plt.title("Actual Factory Material Types vs Predicted Factory Material Types")
plt.xlabel("Actual Factory Material Types")
plt.ylabel("Predicted Factory Material Types")
```
