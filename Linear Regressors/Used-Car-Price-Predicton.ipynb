```python
import pandas as pd
from matplotlib import pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

used_car_price_df = pd.read_csv("Used-Car-Price-Prediction.csv")
used_car_price_df = used_car_price_df.dropna()
```

```python
x = used_car_price_df.drop(columns="price_usd")
y = used_car_price_df["price_usd"]

x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=91)
```

```python
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict([[2, 15400, 300, 145, 3]])
print(f"Model Coeffient = {model.coef_}")
print(f"Model Intercept = {model.intercept_}")
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()

try:
    print(f"Car Price = {y_pred}")
except Exception:
    print(f"Error = {Exception}")    
```

```python
plt.figure(figsize=(11,2))
plt.scatter(y_pred, y_test)
plt.title("Actual Car Prices vs Predicted Car Prices")
plt.xlabel("Actual Car Prices")
plt.ylabel("Predicted Car Prices")
plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]
)
```
