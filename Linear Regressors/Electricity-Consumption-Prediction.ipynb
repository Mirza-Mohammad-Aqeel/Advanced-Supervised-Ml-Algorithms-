Importing & Cleaning the Data

```python
import pandas as pd
from matplotlib import pyplot as plt

from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.model_selection import train_test_split
from sklearn.linear_model  import LinearRegression

electricity_consumption_df = pd.read_csv("Electricity-Consumption-Prediction.csv")
electricity_consumption_df = electricity_consumption_df.dropna()
```

Splitting the Data

```python
X = electricity_consumption_df.drop(columns="monthly_kwh")
Y = electricity_consumption_df['monthly_kwh']

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=30)
```

Creating & Training the Model

```python
model = LinearRegression()
model.fit(x_train, y_train)
```

Making Predictions & Evaluating Accuracy

```python
y_pred = model.predict([[4, 5, 30, 860, 9]])
print(f"Model Coeffients = {model.coef_}")
print(f"Model Intercept = {model.intercept_}")
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()

try:
    print(f"Electricity consumption = {y_pred} kwh")
except Exception:
    print(f"Error = {Exception}")    
```

Visualizing the Model

```python
plt.figure(figsize=(14,3))
plt.scatter(y_pred, y_test)
plt.title("Actual Electricity Consumption Magnitudes vs Predicted Magnitude")
plt.xlabel("Actual Electricity Consumption")
plt.ylabel("Predicted Maginitudes")
plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
)
plt.show()
```
