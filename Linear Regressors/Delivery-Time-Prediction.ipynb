```python
import pandas as pd
from matplotlib import pyplot as plt

from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

delivery_time_df = pd.read_csv("Delivery-Time-Prediction.csv")
delivery_time_df = delivery_time_df.dropna()
```

```python
X = delivery_time_df.drop(columns="delivery_time_min")
Y = delivery_time_df["delivery_time_min"]

x_train, x_test, y_train, y_test = train_test_split(X,Y,test_size=0.2, random_state=12)
```

```python
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict([[3.5, 6, 7, 2, 5]])
print(f"Model Coeffient = {model.coef_}")
print(f"Model Intercept = {model.intercept_}")
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()

try:
    print(f"Delivery time = {y_pred} minutes")
except Exception:
    print(f"Error = {Exception}")    
```

```python
plt.figure(figsize=(14,5))
plt.scatter(y_test, y_pred)
plt.title("Actual Delivery Times vs Predicted Times")
plt.xlabel("Actual Delivery Times")
plt.ylabel("Predicted Times")
plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]
)
plt.show()
```
