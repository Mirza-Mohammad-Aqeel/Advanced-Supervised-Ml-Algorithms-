Importing & Cleaning the Data

```python
import pandas as pd
from matplotlib import pyplot as plt

from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

house_price_df = pd.read_csv("House-Price-Prediction.csv")
house_price_df = house_price_df.dropna()
```

Splitting the Data

```python
X = house_price_df.drop(columns="price")
Y = house_price_df["price"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=112)
```

Creating & Training the Model

```python
model = LinearRegression()
model.fit(x_train, y_train)
```

Making Predictions & Evaluating Accuracy

```python
y_pred = model.predict([[865, 3, 4, 13, 9.8]])
print(f"Model Coeffients = {model.coef_}")
print(f"Model Intercept = {model.intercept_}")
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()

try:
    print(f"House Price = {y_pred}")
except Exception:
    print(f"Error = {Exception}")    
```

Visualizing the Model

```python
plt.figure(figsize=(14,6))
plt.scatter(y_pred, y_test)
plt.title("Actual House Prices vs Predicted Prices")
plt.xlabel("Actual House Prices")
plt.ylabel("Predicted Prices")
plt.plot(
    [y_test.min(),y_test.max()],
    [y_test.min(),y_test.max()]
)
plt.show()
```
