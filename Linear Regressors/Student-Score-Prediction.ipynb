Importing & Cleaning the Data

```python
import pandas as pd
from matplotlib import pyplot as plt

from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

student_score_df = pd.read_csv("Student-Score-Prediction.csv")
student_score_df = student_score_df.dropna()
```

```python
X = student_score_df.drop(columns="final_score")
Y = student_score_df["final_score"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=87)
```

```python
model = LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred = model.predict([[4, 78, 9, 3, 58]])
print(f"Model Coeffient = {model.coef_}")
print(f"Model Intercept = {model.intercept_}")
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()

try:
    print(f"Student Score = {y_pred}")
except Exception:
    print(f"Error = {Exception}")    
```

```python
plt.figure(figsize=(14,3))
plt.scatter(y_pred, y_test)
plt.title("Actual Student Scores vs Predicted Student Scores")
plt.xlabel("Actual Student Scores")
plt.ylabel("Predicted Student Scores")

plt.plot ( 
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    )
plt.show()
```
