Importing & Cleaning the Data

```python
import pandas as pd
from matplotlib import pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

apartment_rent_df = pd.read_csv("Apartment-Rent-Prediction.csv")
apartment_rent_df = apartment_rent_df.dropna()
```

Splitting the Data

```python
X = apartment_rent_df.drop(columns="monthly_rent")
Y = apartment_rent_df["monthly_rent"]

x_train, x_test, y_train, y_test = train_test_split(X,Y,test_size=0.2, random_state=23)
```

Creating & Training the Model

```python
model = LinearRegression()
model.fit(x_train, y_train)
```

Making Prediction & Evaluating Accuracy

```python
y_pred = model.predict([[570, 9, 8 , 15, 10]])
print(f"Model Coeffients = {model.coef_}")
print(f"Model Intercept = {model.intercept_}")
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")

try:
    print(f"Apartment Rent = {y_pred}")
except Exception:
    print(f"Error = {Exception}")    
```

Visualizing the Model

```python
plt.figure(figsize=(13,4))
plt.scatter(y_test, y_pred)
plt.title("Actual Scores vs Predicted Scores")
plt.xlabel("Actual Scores")
plt.ylabel("Predicted Scores")
plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]
)
plt.show()
```
