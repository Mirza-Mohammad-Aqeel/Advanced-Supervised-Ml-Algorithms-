```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

disease_df = pd.read_csv("Agricultural-Crop-Disease-Severity-Prediction.csv")
disease_df = disease_df.dropna()
```

```python
X = disease_df.drop(columns="disease_severity")
Y = disease_df["disease_severity"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=89)
```

```python
Model  = make_pipeline(
    StandardScaler(),
    RandomForestRegressor()
)
params = {
    "n_estimators":[100, 98, 80, 110],
    "max_depth":[4,5,2,3],
    "random_state":[69,78,23,12]
}
grid = GridSearchCV(Model, params, cv=8)
grid.train(x_train, y_train)
```

```python
y_pred = grid.predict([[53.4,29.8,1.9,2.3,45.6,54,25,2,225,34,15]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_absolute_error(y_pred, y_test)}")
print(f"R2 Score = {mean_absolute_error(y_pred, y_test)}")
print()

importance = pd.DataFrame ({
    "Features":X.columns,
    "Importance":RandomForestRegressor.feature_importances_
})
print(f"Importance =")
print(importance.sort_values(by="Importance", ascending=False))
print()
print(f"Crop Disease Severity = {y_pred}")
```

```python
plt.figure(figsize=(15,6))
plt.bar(
    importance["Features"],
    importance["Importance"]
)
plt.title("Random Forest Regressor Feature Importance")
plt.xlabel("Features")
plt.ylabel("Importances")

plt.figure(figsize=(14,3))
plt.scatter(y_test,y_pred)
plt.title("Actual Disease Severities vs Predicted Disease Severities")
plt.xlabel("Actual Disease Severities")
plt.xlabel("Predicted Disease Severities")
plt.plot (
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]
)
plt.show()
```
