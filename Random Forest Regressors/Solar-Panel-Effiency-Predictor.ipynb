```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

panel_df = pd.read_csv("Solar-Panel-Effiency-Prediction.csv")
panel_df = panel_df.dropna()
```

```python
X = panel_df.drop(columns="efficiency_pct")
Y = panel_df["efficiency_pct"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2 ,random_state=21)
```

```python
Model = make_pipeline(
    StandardScaler(),
    RandomForestRegressor()
)
params = {
    "n_estimators":[100,98,102,99],
    "max_depth":[4,7,2,3],
    "random_state":[45,67,89,23]
}
grid = GridSearchCV(Model, params, cv=8)
grid.train(x_train, y_train)
```

```python
y_pred = grid.predict([[890,29,4,50,3,30,13,3,38.7,16.7,21.0]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()
importance = pd.DataFrame ({
    "Feature":X.columns,
    "Importance":RandomForestRegressor.feature_importances_
})
print(f"Importance =")
print(importance.sort_values(by="Importance", ascending=False))
print()
print(f"Effiency Percentage = {y_pred} %")
```

```python
plt.figure(figsize=(16,5))
plt.bar (
    importance["Feature"], importance["Importance"], color="orange", linewidth=4)
plt.title("Random Forest Regressor Feature Importance")
plt.xlabel("Feature")
plt.ylabel("Importance")

plt.figure(figsize=(15,3))
plt.scatter(y_test, y_pred)
plt.title("Actual Effiency Percentages vs Predicted Satellite Effiency Percentages")
plt.xlabel("Actual Effiency Percentages")
plt.ylabel("Predicted Effiency Percentages")
plt.plot (
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]
)
plt.show()
```
