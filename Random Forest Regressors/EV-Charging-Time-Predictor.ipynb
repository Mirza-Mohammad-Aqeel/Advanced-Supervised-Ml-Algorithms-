```python
import pandas as pd
from matplotlib import pyplot as plt
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

charging_time_df = pd.read_csv("EV-Charging-Time-Prediction.csv")
charging_time_df = charging_time_df.dropna()
```

```python
X = charging_time_df.drop(columns="charging_time_min")
Y = charging_time_df["charging_time_min"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=34)
```

```python
Model = make_pipeline(
    StandardScaler(),
    RandomForestRegressor()
)
params = {
    "n_estimators":[100,80,120,110],
    "max_depth":[4,6,2,3],
    "random_state":[89,76,90,45]
}
grid = GridSearchCV(Model, params, cv=5)
grid.train(x_train, y_train)
```

```python
y_pred = grid.predict([[31,70,150,55,33,94,2,5.3,440,28]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print(f"Best Params = {grid.best_params_}")
print()
importance = pd.DataFrame ({
    "Features":X.columns,
    "Importance":RandomForestRegressor.feature_importances_
})
print(f"Importance =")
print(importance.sort_values(by="Importance", ascending=False))
print()
print(f"Charging Time = {y_pred} min")
```

```python
plt.figure(figsize=(15,6))
plt.bar (
    importance["Features"], importance["Importance"]
)
plt.title("Random Forest Regressor Feature Importance")
plt.xlabel("Feature")
plt.ylabel("Importance")

plt.figure(figsize=(14,3))
plt.scatter(y_test, y_pred)
plt.title("Actual Charging Times vs Predicted Charging Times")
plt.xlabel("Actual Charging Times")
plt.ylabel("Predicted Charging Times")
plt.plot (
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]
)
plt.show()
```
