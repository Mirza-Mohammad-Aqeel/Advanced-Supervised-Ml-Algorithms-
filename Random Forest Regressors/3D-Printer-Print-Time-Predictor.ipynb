```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.ensemble import RandomForestRegressor

printer_df = pd.read_csv("3D-Printer-Print-Time-Prediction")
printer_df = printer_df.dropna()
```

```python
X = printer_df.drop(columns="print_time_min")
Y = printer_df["print_time_min"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=56)
```

```python
Model = make_pipeline(
    StandardScaler(),
    RandomForestRegressor()
)
params = {
    "n_estimators":[95, 105, 100, 80],
    "max_depth":[4,2,3,1],
    "random_state":[42, 55, 67, 12],

}
grid = GridSearchCV(Model, params, cv=8)
grid.train(x_train, y_train)
```

```python
y_pred = grid.predict([[685,0.16,49.8,40,0.4,49,4,122.4,1.27,562]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_absolute_error(y_pred, y_test)}")
print(f"R2 Score = {mean_absolute_error(y_pred, y_test)}")
print()
importance = pd.DataFrame ({
    "Features":X.columns,
    "Importance":RandomForestRegressor.feature_importances_
})
print(f"Importance =")
print(importance.sort_values(by="Importance", ascending=False))
print()
print(f"Printing Time = {y_pred} min")
```

```python
plt.figure(figsize=(17,8))
plt.bar (
    importance["Feature"], importance["Importance"]
)
plt.title("Random Forest Regressor Feature Importance")
plt.xlabel("Feature")
plt.ylabel("Importance")

plt.figure(figsize=(14,5))
plt.scatter(y_test, y_pred)
plt.title("Actual Printing Times vs Predicted Printing Times")
plt.xlabel("Actual Printing Times")
plt.ylabel("Predicted Printing Times")
plt.plot (
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()])
plt.show()
```
