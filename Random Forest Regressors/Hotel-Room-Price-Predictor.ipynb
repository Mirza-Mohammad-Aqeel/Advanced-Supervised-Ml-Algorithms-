```python
import pandas as pd
from matplotlib import pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.ensemble import RandomForestRegressor

room_price_df = pd.read_csv("Hotel-Room-Price-Prediction.csv")
room_price_df = room_price_df.dropna()
```

```python
X = room_price_df.drop(columns="room_price")
Y = room_price_df["room_price"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=63)
```

```python
Model = make_pipeline(
    StandardScaler(),
    RandomForestRegressor()
)
params = {
    "n_estimators":[140,120,100,150],
    "max_depth":[4,6,3,2],
    "random_state":[34,67,45,89]
}
grid = GridSearchCV(Model, params, cv=6)
grid.train(x_train, y_train)
```

```python
y_pred = grid.predict([[1.5,38,5,93,4,90,1,3,15,275]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print(f"Best Params = {grid.best_params_}")
print()
importance = pd.DataFrame ({
    "Feature":X.columns,
    "Importance":RandomForestRegressor.feature_importances_
})
print(f"Importances =")
print(importance.sort_values(by="Importance", ascending=False))
print()
print(f"Hotel Room Prices = {y_pred} $")
```

```python
plt.figure(figsize=(14,5))
plt.bar (
    importance["Feature"], importance["Importance"]
)
plt.title("Random Forest Regressor Feature Importance")
plt.xlabel("Feature")
plt.ylabel("Importance")

plt.figure(figsize=(15,6))
plt.scatter(y_test, y_pred)
plt.title("Actual Hotel Room Prices vs Predicted Hotel Rooms Prices")
plt.xlabel("Actual Hotel Room Prices")
plt.ylabel("Predicted Hotel Room Prices")
plt.plot (
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]
)
plt.show()
```
