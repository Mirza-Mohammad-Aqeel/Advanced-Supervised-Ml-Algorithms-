```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.ensemble import RandomForestRegressor

traffic_df = pd.read_csv("Airport-Passenger-Traffic-Prediction.csv")
traffic_df = traffic_df.dropna()
```

```python
X = traffic_df.drop(columns="passengers_per_day")
Y = traffic_df["passengers_per_day"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2 , random_state=23)
```

```python
Model = make_pipeline(
    StandardScaler(),
    RandomForestRegressor()
)
params = {
    "n_estimators":[100,98,103,95],
    "max_depth":[4,5,3,2],
    "random_state":[34,56,44,23]
}
grid = GridSearchCV(Model, params, cv=4)
grid.train(x_train, y_train)
```

```python
y_pred = grid.predict([[3.1,145,58,69,34.2,21,39,65,151,12980]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {mean_absolute_error(y_pred, y_test)}")
print()

importance = pd.DataFrame ({
    "Features":X.columns,
    "Importance":RandomForestRegressor.feature_importances_
})
print(f"Importance =")
print(importance.sort_values(by="Importance", ascending=False))
print()
print(f"Passenger Traffic = {y_pred} passengers per Day")
```

```python
plt.figure(figsize=(16,5))
plt.bar (
    importance["Features"],importance["Importance"]
)
plt.title("Random Forest Regressor Feature Importance")
plt.xlabel("Feature")
plt.ylabel("Importance")

plt.figure(figsize=(14,5))
plt.scatter(y_test,y_pred)
plt.title("Actual Passenger Traffic Magnitudes vs Predicted Passenger Traffic Magnitudes")
plt.xlabel("Actual Passenger Traffic Magnitudes")
plt.ylabel("Predicted Passenger Traffic Magnitudes")
plt.plot (
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]
)
plt.show()
```
