```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline

exam_score_df =  pd.read_csv("Exam-Score-Prediction.csv")
exam_score_df = exam_score_df.dropna()
```

```python
X = exam_score_df.drop(columns="exam_score")
Y = exam_score_df["exam_score"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=34)
```

```python
model = make_pipeline(
    PolynomialFeatures(degree=5)
    LinearRegression()
)
model.fit(x_train, y_train)
```

```python
y_pred = model.predict([[2,4,79,81,34]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()

print(f'Predicted Exam Score = {y_pred}')
X_curve = np.linspace(X.columns.min(), X.columns.max()).reshape(-1,1)
y_curve = model.predict(X_curve)
```

```python
plt.figure(figsize=(10,9))
plt.scatter(X_curve, y_curve)
plt.title("Polynomial Regression")
plt.xlabel(X.columns)
plt.ylabel("Predicted Exam Scores")
plt.show()
```
