```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline

revenue_prediction = pd.read_csv("Revenue-Prediction.csv")
revenue_prediction = revenue_prediction.dropna()
```

```python
X = revenue_prediction.drop(columns="revenue_k")
Y = revenue_prediction["revenue_k"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=51)
```

```python
model = make_pipeline(
    PolynomialFeatures(degree=2),
    LinearRegression()
)
model.fit(x_train, y_train)
```

```python
y_pred = model.predict([[2,13,4,5]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MsE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()
print(f"Revenue = {y_pred}k Dollars")
x_curve = np.linspace(X.columns.min(), X.columns.max()).reshape(-1,1)
y_curve = model.predict(x_curve)
```

```python
plt.figure(figsize=(13,2))
plt.scatter(x_curve, y_curve)
plt.title("Polynomial Regression")
plt.xlabel(X.columns)
plt.ylabel("Predicted Revenues")
plt.show()
```
