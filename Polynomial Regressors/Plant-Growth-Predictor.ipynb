```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.model_selection import train_test_split

plant_growth_df = pd.read_csv("Plant-Growth-Prediction.csv")
plant_growth_df = plant_growth_df.dropna()
```

```python
X = plant_growth_df.drop(columns="height_cm")
Y = plant_growth_df["height_cm"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=46)
```

```python
model = make_pipeline(
    PolynomialFeatures(degree=4)
    LinearRegression()
)
model.fit(x_train, y_train)
```

```python
y_pred = model.predict([[3,5,156,34,4]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()

print(f"Predicted Plant Growth = {y_pred} cm")
x_curve = np.linspace(X.columns.min(), X.columns.max()).reshape(-1,1)
y_curve = model.predict(x_curve)
```

```python
plt.figure(figsize=(13,2))
plt.scatter(x_curve, y_curve)
plt.title("Polynomial Regression")
plt.xlabel(X.columns)
plt.ylabel("Predicted Plant Growths")
plt.show()
```
