```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline

productivity_df = pd.read_csv("Productivity-Score-Prediction.csv")
productivity_df = productivity_df.dropna()
```

```python
X = productivity_df.drop(columns="productivity_score")
Y = productivity_df["productivity_score"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=51)
```

```python
model = make_pipeline(
    PolynomialFeatures(degree=2)
    LinearRegression()
)
model.fit(x_train, y_train)
```

```python
y_pred = model.predict([[2,8,9,1]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MsE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()
print(f"Productivity Score = {y_pred}")
x_curve = np.linspace(X.columns.min(), X.columns.max()).reshape(-1,1)
y_curve = model.predict(x_curve)
```

```python
plt.figure(figsize=(13,2))
plt.scatter(x_curve, y_curve)
plt.title("Polynomial Regression")
plt.xlabel(X.columns)
plt.ylabel("Predicted Productivity Scores")
plt.show()
```
