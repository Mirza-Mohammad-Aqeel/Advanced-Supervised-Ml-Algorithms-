```python
import pandas as pd
from matplotlib import pyplot as plt
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.linear_model import LinearRegression

calories_burned = pd.read_csv("Calories-Burned-Prediction.csv")
calories_burned = calories_burned.dropna()
```

```python
X = calories_burned.drop(columns="calories_burned")
Y = calories_burned["calories_burned"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=11)
```

```python
model = make_pipeline (
    PolynomialFeatures(degree=2)
    LinearRegression()
)
model.fit(x_train, y_train)
```

```python
y_pred = model.predict([[14,5,12,89,67]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
X_curve = np.linspace(X.columns.min(), X.columns.max(), 200).reshape(-1,1)
Y_curve = model.predict(X_curve)
print()
try:
    print(f"Calories burned = {y_pred}")
except Exception:
    print(f"Error = {Exception}")    
```

```python
plt.figure(figsize=(13,4))
plt.scatter(X_curve,Y_curve)
plt.title("Polynomial Regression")
plt.xlabel(X.columns)
plt.ylabel("Calories Burned")
plt.show()
```
