```python
import numpy as np
import pandas as pd
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.model_selection import train_test_split
from matplotlib import pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LinearRegression

employee_salary_df = pd.read_csv("Employee-Salary-Prediction.csv")
employee_salary_df = employee_salary_df.dropna()
```

```python
X = employee_salary_df.drop(columns="salary_k")
Y = employee_salary_df["salary_k"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=21)
```

```python
model = make_pipeline (
    PolynomialFeatures(degree=4)
    LinearRegression()
)
model.fit(x_train,y_train)
```

```python
y_pred = model.predict([[1,19,70,14,23]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()

print(f"Predicted Salary = {y_pred}")
X_curve = np.linspace(X.columns.min(), X.columns.max(), 200).reshape(-1,1)
y_curve = model.predict(X_curve)
```

```python
plt.figure(figsize=(11,2))
plt.scatter(X_curve, y_curve)
plt.title("Polynomial Regression")
plt.xlabel(X.columns)
plt.ylabel("Predicted Salary")
plt.show()
```
