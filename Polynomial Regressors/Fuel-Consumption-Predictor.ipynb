```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.model_selection import train_test_split

fuel_consumption_df = pd.read_csv("Fuel-Consumption-Prediction.csv")
fuel_consumption_df = fuel_consumption_df.dropna()
```

```python
X = fuel_consumption_df.drop(columns="fuel_l_per_100km")
Y = fuel_consumption_df["fuel_l_per_100km"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=46)
```

```python
model = make_pipeline(
    PolynomialFeatures(degree=2)
    LinearRegression()
)
model.fit(x_train, y_train)
```

```python
y_pred = model.predict([[29, 1540, 1300, 4]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()

print(f"Predicted Fuel Consumption = {y_pred} litres per km")
x_curve = np.linspace(X.columns.min(), X.columns.max()).reshape(-1,1)
y_curve = model.predict(x_curve)
```

```python
plt.figure(figsize=(9,12))
plt.scatter(x_curve, y_curve)
plt.title("Polynomial Regression")
plt.xlabel(X.columns)
plt.ylabel("Predicted Fuel Consumption Magnitudes")
plt.show()
```
