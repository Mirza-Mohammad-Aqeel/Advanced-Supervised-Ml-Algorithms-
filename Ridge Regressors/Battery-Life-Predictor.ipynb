```python
import pandas as pd
from matplotlib import pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import ridge_regression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.pipeline import make_pipeline 
from sklearn.preprocessing import StandardScaler

battery_life_df = pd.read_csv("Battery-Life-Prediction.csv")
battery_life_df = battery_life_df.dropna()
```

```python
X = battery_life_df.drop(columns="battery_life_hours")
Y = battery_life_df["battery_life_hours"]

x_train, x_test , y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=91)
```

```python
model = make_pipeline (
    ridge_regression(alpha=1.9),
    StandardScaler
)
model.fit(x_train, y_train)
```

```python
y_pred = model.predict([[46,12,9.5,2, 8, 9 , 2.4, 5400]])
print(f"Model Coeffient = {model.coef_}")
print(f"Model Intercept = {model.intercept_}")
print(f"MAE = {mean_absolute_error(y_pred,y_test)}")
print(f"MSE = {mean_absolute_error(y_pred,y_test)}")
print(f"R2 Score = {mean_absolute_error(y_pred,y_test)}")
print()
print(f"Battery Life = {y_pred} hours")
```

```python
plt.figure(figsize=(10,9))
plt.scatter(y_test, y_pred)
plt.title("Actual Battery Hours vs Predicted Battery Hours")
plt.xlabel("Actual Battery Hours")
plt.ylabel("Predicted Battery Hours")
plt.plot (
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
)
plt.show()
```
