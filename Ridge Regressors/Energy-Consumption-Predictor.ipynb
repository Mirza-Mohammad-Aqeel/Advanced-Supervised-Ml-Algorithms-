```python
import  pandas as pd
from matplotlib import pyplot as plt
from sklearn.linear_model import ridge_regression
from sklearn.metrics import mean_squared_error, mean_absolute_error , r2_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

energy_consumption_df =pd.read_csv("Manufacturing-Machine-Energy-Consumption-Prediction.csv")
energy_consumption_df = energy_consumption_df.dropna()
```

```python
X = energy_consumption_df.drop(columns="energy_kwh")
Y = energy_consumption_df["energy_kwh"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.3, random_state=14)
```

```python
model = make_pipeline(
    StandardScaler(),
    ridge_regression(alpha=1.3)
)
model.fit(x_train, y_train)
```

```python
y_pred = model.predict([[35, 1320, 79, 5.6, 2.3, 9, 180 , 20]])
print(f"Model Coeffient = {model.coef_}")
print(f"Model Intercept = {model.intercept_}")
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()
print(f"Energy Consumption = {y_pred} kwh")
```

```python
plt.figure(figsize=(13,4))
plt.scatter(y_test, y_pred)
plt.title("Actual Energy Consumptions vs Predicted Consumptions")
plt.xlabel("Actual Energy Consumptions")
plt.ylabel("Predicted Consumptions")
plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]

)
plt.show()
```
