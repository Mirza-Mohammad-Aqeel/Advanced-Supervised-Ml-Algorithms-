```python
import pandas as pd
from matplotlib import pyplot as plt
from sklearn.linear_model import ridge_regression
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.model_selection import train_test_split

grid_load_df = pd.read_csv("Power-Grid-Load-Prediction.csv")
grid_load_df = grid_load_df.dropna()
```

```python
X = grid_load_df.drop(columns="current_load_mw")
Y = grid_load_df["current_load_mw"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=61)
```

```python
model = make_pipeline(
    StandardScaler(),
    ridge_regression(alpha=3.2)
)
model.fit(x_train, y_train)
```

```python
y_pred = model.predict([[19, 43, 5.6, 12300, 39, 4.2, 83, 87]])
print(f"Model Coeffient = {model.coef_}")
print(f"Model Intercept = {model.intercept_}")
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()
print(f"Grid Load = {y_pred} mw")
```

```python
plt.figure(figsize=(15,4))
plt.scatter(y_test, y_pred)
plt.title("Actual Grid Loads vs Predicted Grid Loads")
plt.xlabel("Actual Grid Loads")
plt.ylabel("Predicted Grid Loads")
plt.plot (
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]
)
plt.show()
```
