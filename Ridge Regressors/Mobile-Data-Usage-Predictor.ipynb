```python
import pandas as pd
from matplotlib import pyplot as plt
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
from sklearn.linear_model import ridge_regression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

data_usage_df = pd.read_csv("Mobile-Data-Usage-Prediction.csv")
data_usage_df = data_usage_df.dropna()
```

```python
X = data_usage_df.drop(columns="data_usage_gb")
Y = data_usage_df["data_usage_gb"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=34)
```

```python
model = make_pipeline(
    StandardScaler(),
    ridge_regression(alpha=2.3)
)
model.fit(x_train, y_train)
```

```python
y_pred = model.predict([[3.4, 1.5, 0.9, 8.7, 1.4, 3.5, 41, 9.6]])
print(f"Model Coeffeint = {model.coef_}")
print(f"Model Intercept = {model.intercept_}")
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()
print(f"Data Used = {y_pred} GB")
```

```python
plt.figure(figsize=(14,5))
plt.scatter(y_test, y_pred)
plt.title("Actual Data Usages vs Predicted Usages")
plt.xlabel("Actual Data Usages")
plt.ylabel("Predicted Usages")
plt.plot (
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]
)
plt.show()
```
