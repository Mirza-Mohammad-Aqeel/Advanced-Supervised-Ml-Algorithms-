```python
import pandas as pd
from matplotlib import pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import ridge_regression

patient_stay_df = pd.read_csv("Hospital-Patient-StayLength-Prediction.csv")
patient_stay_df = patient_stay_df.dropna()
```

```python
X = patient_stay_df.drop(columns="stay_days")
Y = patient_stay_df["stay_days"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=12)
```

```python
model = make_pipeline(
    StandardScaler(),
    ridge_regression(alpha=2.2)
)
model.fit(x_train, y_train)
```

```python
y_pred = model.predict([[24, 34, 119, 77.8, 4, 3, 0.9, 2, 12]])
print(f"Model Coeffient = {model.coef_}")
print(f"Model Intercept = {model.intercept_}")
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()

print(f"Patient Stay = {y_pred} days")
```

```python
plt.figure(figsize=(15,4))
plt.scatter(y_test, y_pred)
plt.title("Actual Patient Stays vs Predicted Stays")
plt.xlabel("Actual Patient Stays ( Days )")
plt.ylabel("Predicted Stays")
plt.plot (

    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]

)
plt.show()
```
