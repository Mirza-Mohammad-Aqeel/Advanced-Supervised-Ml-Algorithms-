```python
import pandas as pd
from matplotlib import pyplot as plt
from sklearn.linear_model import ridge_regression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

taxi_fare_df = pd.read_csv("Taxi-Fare-Prediction.csv")
taxi_fare_df = taxi_fare_df.dropna()
```

```python
X = taxi_fare_df.drop(columns="fare")
Y = taxi_fare_df["fare"]

x_train, x_test, y_train, y_test =train_test_split(X,Y, test_size=0.2, random_state=82)
```

```python
model = make_pipeline(
    StandardScaler(),
    ridge_regression(alpha=1.3)
)
model.fit(x_train, y_train)
```

```python
y_pred = model.predict([[3.4, 4.5, 1.9, 28, 3.9, 6.4, 2]])
print(f"Model Coeffient = {model.coef_}")
print(f"Model Intercept = {model.intercept_}")
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()
print(f"Taxi Fare = {y_pred}")
```

```python
plt.figure(figsize=(15,4))
plt.scatter(y_test, y_pred)
plt.title("Actual Taxi Fares vs Predicted Fares")
plt.xlabel("Actual Taxi Fares")
plt.ylabel("Predicted Fares")
plt.plot (
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]
)
plt.show()
```
