```python
import pandas as pd
from matplotlib import pyplot as plt
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.model_selection import  train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model  import ridge_regression

water_consumption_df = pd.read_csv("Water-Consumption-Prediction.csv")
water_consumption_df = water_consumption_df.dropna()
```

```python
X = water_consumption_df.drop(columns="daily_consumption_m3")
Y = water_consumption_df["daily_consumption_m3"]

x_train, x_test, y_train, y_test= train_test_split(X,Y, test_size=0.2, random_state=25)
```

```python
model = make_pipeline(
    StandardScaler(),
    ridge_regression(alpha=2.6)
)
model.fit(x_train, y_train)
```

```python
y_pred = model.predict([[23.4, 815, 8, 3.4, 23, 45, 47, 1.9]])
print(f"Model Coeffient = {model.coef_}")
print(f"Model Intercept = {model.intercept_}")
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()
print(f"Water Consumption = {y_pred}")
```

```python
plt.figure(figsize=(15,7))
plt.scatter(y_test, y_pred)
plt.title("Actual Water Consumptions vs Predicted Consumptions")
plt.xlabel("Actual Water Consumptions")
plt.ylabel("Predicted Consumptions")
plt.plot (
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]
)
plt.show()
```
