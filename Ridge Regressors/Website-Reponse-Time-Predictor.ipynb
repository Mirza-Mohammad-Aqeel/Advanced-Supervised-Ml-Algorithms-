```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import ridge_regression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

website_reponse_time_df = pd.read_csv("Website-Response-Time-Prediction.csv")
website_reponse_time_df = website_reponse_time_df.dropna()
```

```python
X = website_reponse_time_df.drop(columns="response_time_ms")
Y = website_reponse_time_df["response_time_ms"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=62)
```

```python
model = make_pipeline(
    StandardScaler(),
    ridge_regression(alpha=2.3)
)
model.fit(x_train, y_train)
```

```python
y_pred = model.predict([[145.6, 23.4, 31.9, 275, 422, 19.27, 2.3]])
print(f"Model Coeffient = {model.coef_}")
print(f"Model Intercept = {model.intercept_}")
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()
print(f"Website Response Time = {y_pred} ms")
```

```python
plt.figure(figsize=(16,5))
plt.scatter(y_test, y_pred)
plt.title("Actual Loading Times vs Predicted Loading Times")
plt.xlabel("Actual Loading Times")
plt.ylabel("Predicted Loading Times")
plt.plot (
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]
)
plt.show()
```
