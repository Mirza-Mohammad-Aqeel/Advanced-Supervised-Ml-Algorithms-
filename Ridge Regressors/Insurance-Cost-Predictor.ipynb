```python
import pandas as pd
from matplotlib import pyplot as plt
from sklearn.linear_model import ridge_regression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

insurance_cost_df = pd.read_csv("Medical-Insurance-Cost-Prediction.csv")
insurance_cost_df = insurance_cost_df.dropna()
```

```python
X = insurance_cost_df.drop(columns="target_cost")
Y = insurance_cost_df["target_cost"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=32)
```

```python
model = make_pipeline(
    StandardScaler(),
    ridge_regression(alpha=2.3)
)
model.fit(x_train, y_train)
```

```python
y_pred = model.predict([[ 23, 45, 6, 23500, 7, 3, 2]])
print(f"Model Coeffient = {model.coef_}")
print(f"Model Intercept = {model.intercept_}")
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_absolute_error(y_pred, y_test)}")
print(f"R2 Score = {mean_absolute_error(y_pred, y_test)}")
print()

try:
    print(f"Insurance Cost = {y_pred} $")
except Exception:
    print(f"Error = {Exception}")    
```

```python
plt.figure(figsize=(12,3))
plt.scatter(y_test, y_pred)
plt.title("Actual Insurance Costs vs Predicted Costs")
plt.xlabel("Actual Insurance Costs")
plt.ylabel("Predicted Costs")
plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]
)
plt.show()
```
