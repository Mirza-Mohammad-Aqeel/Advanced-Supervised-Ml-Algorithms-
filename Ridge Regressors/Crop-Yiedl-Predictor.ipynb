```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
from sklearn.linear_model import ridge_regression
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

crop_yield_df = pd.read_csv("Crop-Yield-Prediction.csv")
crop_yield_df = crop_yield_df.dropna()
```

```python
X = crop_yield_df.drop(columns="yield_tons")
Y = crop_yield_df["yield_tons"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=11)
```

```python
model = make_pipeline(
    StandardScaler(),
    ridge_regression(alpha=1.8)
)
model.fit(x_train, y_train)
```

```python
y_pred = model.predict([[450, 29, 3, 12, 39, 41, 17, 9, 8.2]])
print(f"Model Coeffients = {model.coef_}")
print(f"Model Intercept = {model.intercept}")
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()
print(f"Crop yields = {y_pred} tons")
```

```python
plt.figure(figsize=(12,8))
plt.scatter(y_test, y_pred)
plt.title("Actual Crop Yields vs Predicted Crop Yields")
plt.xlabel("Actual Crop Yields")
plt.ylabel("Predicted Crop Yields")
plt.plot (
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]
)
plt.show()
```
