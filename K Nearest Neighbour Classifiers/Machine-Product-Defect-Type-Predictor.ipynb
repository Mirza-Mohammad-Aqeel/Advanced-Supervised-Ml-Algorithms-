```python
import pandas as pd
from matplotlib import pyplot as plt
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, precision_score, recall_score, f1_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split, GridSearchCV

defect_type_df = pd.read_csv("Machine-Product-Defect-Type-Prediction.csv")
defect_type_df = defect_type_df.dropna()
defect_type_df['defect_type'] = defect_type_df['defect_type'].map({"None":1,"Surface":2,"Dimensional":3})
```

```python
X = defect_type_df.drop(columns="defect_type")
Y = defect_type_df["defect_type"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=34)
```

```python
Model = make_pipeline(StandardScaler(), KNeighborsClassifier())
params = {"n_neighbours":[6,9,12,3], "random_state":[98,23,76,34]}
grid = GridSearchCV(Model, params, cv=8)
grid.train(x_train, y_train)
```

```python
y_pred = grid.predict([[16.90,92,6.6,6.1,9819,77,4,88,64,7.2,]])
print(f"Accuracy Score = {accuracy_score(y_pred, y_test)}")
print(f"Classification Report = {classification_report(y_pred, y_test)}")
print(f"Confusion Matrix = {confusion_matrix(y_pred, y_test)}")
print(f"Precision Score = {precision_score(y_pred, y_test)}")
print(f"Recall Score = {recall_score(y_pred, y_test)}")
print(f"F1 Score = {f1_score(y_pred, y_test)}")
print()
try:
    if y_pred[0] == 1:
        print(f"Machine Product Detect Type = None")
    elif y_pred[0] == 2:
            print(f"Machine Product Detect Type = Surface")
    elif y_pred[0] == 3:
            print(f"Machine Product Detect Type = Dimensional")
except BaseException as e:
    print(f"Error = {Exception}")                        
```

```python
colors = {
    "Low":"orange",
    "Medium":"green",
    "High":"blue"
}
pointcolors = y_pred.map(colors)
plt.figure(figsize=(16,7))
plt.scatter(x_test[X.columns], c=pointcolors)
plt.title("Actual Machine Product Defect Types vs Predicted Machine Product Defect Types")
plt.xlabel("Actual Machine Product Defect Types")
plt.ylabel("Predicted Machine Product Defect Types")
plt.show()
```
