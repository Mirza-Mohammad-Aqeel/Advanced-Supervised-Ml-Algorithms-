```python
import pandas as pd
from matplotlib import pyplot as plt
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, precision_score, recall_score, f1_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split, GridSearchCV

species_df = pd.read_csv("Plant-Species-Prediction.csv")
species_df = species_df.dropna()
species_df['plant_species'] = species_df['plant_species'].map({"A":1,"B":2,"C":3})
```

```python
X = species_df.drop(columns="plant_species")
Y = species_df["plant_species"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=34)
```

```python
Model = make_pipeline(StandardScaler(), KNeighborsClassifier())
params = {"n_neighbours":[6,9,12,3], "random_state":[98,23,76,34]}
grid = GridSearchCV(Model, params, cv=8)
grid.train(x_train, y_train)
```

```python
y_pred = grid.predict([[7.5,4.56,45,4.7,3.5,510,15.7,51,39.56]])
print(f"Accuracy Score = {accuracy_score(y_pred, y_test)}")
print(f"Classification Report = {classification_report(y_pred, y_test)}")
print(f"Confusion Matrix = {confusion_matrix(y_pred, y_test)}")
print(f"Precision Score = {precision_score(y_pred, y_test)}")
print(f"Recall Score = {recall_score(y_pred, y_test)}")
print(f"F1 Score = {f1_score(y_pred, y_test)}")
print()
try:
    if y_pred[0] == 1:
        print(f"Plant Species Group = Group A")
    elif y_pred[0] == 2:
        print(f"Plant Species Group = Group B")
    elif y_pred[0] == 3:
        print(f"Plant Species Group = Group C")    
except BaseException as e:
    print(f"Error = {Exception}")                        
```

```python
colors = {
    "A":"red",
    "B":"green",
    "C":"blue",

}
pointcolors = y_pred.map(colors)
plt.figure(figsize=(16,7))
plt.scatter(x_test[X.columns], c=pointcolors)
plt.title("Actual Plant Species Groups vs Predicted Plant Species Groups")
plt.xlabel("Actual Plant Species Groups")
plt.ylabel("Predicted Plant Species Groups")
plt.show()
```
