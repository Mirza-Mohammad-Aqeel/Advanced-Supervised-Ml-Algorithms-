```python
import pandas as pd
from matplotlib import pyplot as plt
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, precision_score, recall_score, f1_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split, GridSearchCV

student_df = pd.read_csv("E-Learning-Student-Engagement-Prediction.csv")
student_df = student_df.dropna()
student_df['engagement'] = student_df['engagement'].map({"Low":1,"Medium":2,"High":3})
```

```python
X = student_df.drop(columns="engagement")
Y = student_df["engagement"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=34)
```

```python
Model = make_pipeline(StandardScaler(), KNeighborsClassifier())
params = {"n_neighbours":[6,9,12,3], "random_state":[98,23,76,34]}
grid = GridSearchCV(Model, params, cv=8)
grid.train(x_train, y_train)
```

```python
y_pred = grid.predict([[198.45,8,23.56,6,94,8.6,85,89,1,-24]])
print(f"Accuracy Score = {accuracy_score(y_pred, y_test)}")
print(f"Classification Report = {classification_report(y_pred, y_test)}")
print(f"Confusion Matrix = {confusion_matrix(y_pred, y_test)}")
print(f"Precision Score = {precision_score(y_pred, y_test)}")
print(f"Recall Score = {recall_score(y_pred, y_test)}")
print(f"F1 Score = {f1_score(y_pred, y_test)}")
print()
try:
    if y_pred[0] == 1:
        print(f"E Learning Student Engagement Status = Low")
    elif y_pred[0] == 2:
            print(f"E Learning Student Engagement Status = Medium")
    elif y_pred[0] == 3:
            print(f"E Learning Student Engagement Status = High")
except BaseException as e:
    print(f"Error = {Exception}")                        
```

```python
colors = {
    "Low":"orange",
    "Medium":"green",
    "High":"blue"
}
pointcolors = y_pred.map(colors)
plt.figure(figsize=(16,7))
plt.scatter(x_test[X.columns], c=pointcolors)
plt.title("Actual E-Learning Student Engagement Statuses vs Predicted E-Learning Student Engagement Statuses")
plt.xlabel("Actual E-Learning Student Engagement Statuses")
plt.ylabel("Predicted E-Learning Student Engagement Statuses")
plt.show()
```
