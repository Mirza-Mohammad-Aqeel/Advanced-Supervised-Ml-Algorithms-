```python
import pandas as pd
from matplotlib import pyplot as plt
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, precision_score, recall_score, f1_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split, GridSearchCV

music_genre_df = pd.read_csv("Music-Genre-Prediction.csv")
music_genre_df = music_genre_df.dropna()
music_genre_df['genre'] = music_genre_df['genre'].map({"Pop":1,"Classical":2,"Electronic":3,"HipHop":4,"Rock":5})
```

```python
X = music_genre_df.drop(columns="genre")
Y = music_genre_df["genre"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=34)
```

```python
Model = make_pipeline(StandardScaler(), KNeighborsClassifier())
params = {"n_neighbours":[6,9,12,3], "random_state":[98,23,76,34]}
grid = GridSearchCV(Model, params, cv=8)
grid.train(x_train, y_train)
```

```python
y_pred = grid.predict([[12.6,81,734,8,21,6,-5.8,21.5,80,9]])
print(f"Accuracy Score = {accuracy_score(y_pred, y_test)}")
print(f"Classification Report = {classification_report(y_pred, y_test)}")
print(f"Confusion Matrix = {confusion_matrix(y_pred, y_test)}")
print(f"Precision Score = {precision_score(y_pred, y_test)}")
print(f"Recall Score = {recall_score(y_pred, y_test)}")
print(f"F1 Score = {f1_score(y_pred, y_test)}")
print()
try:
    if y_pred[0] == 1:
        print(f"Music Genre = Pop")
    elif y_pred[0] == 2:
        print(f"Music Genre = Classical")
    elif y_pred[0] == 3:
        print(f"Music Genre = Electronic")
    elif y_pred[0] == 4:
        print(f"Music Genre = Hip Hop")
    elif y_pred[0] == 5:
        print(f"Music Genre = Rock")                            
except BaseException as e:
    print(f"Error = {Exception}")                        
```

```python
colors = {
    "Low":"orange",
    "Medium":"green",
    "High":"blue"
}
pointcolors = y_pred.map(colors)
plt.figure(figsize=(16,7))
plt.scatter(x_test[X.columns], c=pointcolors)
plt.title("Actual Music Genres vs Predicted Music Genres")
plt.xlabel("Actual Music Genres")
plt.ylabel("Predicted Music Genres")
plt.show()
```
