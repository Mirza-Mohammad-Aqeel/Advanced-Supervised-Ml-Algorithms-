```python
import pandas as pd
from matplotlib import pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, f1_score, recall_score,precision_score

house_type_df = pd.read_csv("House-Type-Prediction.csv")
house_type_df = house_type_df.dropna()
house_type_df['large_house'] = house_type_df['large_house'].map({"Yes":1,"No":0})
```

```python
X = house_type_df.drop(columns="large_house")
Y = house_type_df["large_house"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=43)
```

```python
model = LogisticRegression()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict([[680, 4, 5, 2, 27, 340]])
print(f"Accuracy Score = {accuracy_score(y_pred, y_test)}")
print(f"Classification Report = {classification_report(y_pred, y_test)}")
print(f"Confusion Matrix = {confusion_matrix(y_pred, y_test)}")
print(f"F1 Score = {f1_score(y_pred, y_test)}")
print(f"Precision Score = {precision_score(y_pred, y_test)}")
print(f"Recall Score = {recall_score(y_pred, y_test)}")
print()

try:
    if y_pred[0] == 1:
        print(f"Is your House large = Yes")
    elif y_pred[0] == 0:
        print(f"Is your House large = No")
except Exception:
    print(f"Error = {Exception}")            
```

```python
plt.figure(figsize=(13,2))
plt.scatter (
    x_test["area_sqft"],
    x_test["bedrooms"],
    x_test["bathrooms"],
    x_test["garage_spaces"],
    x_test["age_years"],
    x_test["garden_area_sqft"],
    c = y_pred
)
plt.title("Logisic Regression")
plt.xlabel(X.columns)
plt.ylabel("Large House")
plt.show()
```
