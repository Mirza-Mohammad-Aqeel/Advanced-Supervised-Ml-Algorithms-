```python
import pandas as pd
from matplotlib import pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, f1_score, precision_score, recall_score

card_df = pd.read_csv("Credit-Card-Default.csv")
card_df = card_df.dropna()
card_df['defaulted'] = card_df['defaulted'].map({"Yes":1,"No":0})
```

```python
X = card_df.drop(columns="defaulted")
Y = card_df["defaulted"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=43)
```

```python
model = LogisticRegression()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict([[26, 578, 13,1, 4]])
print(f"Accuracy Score = {accuracy_score(y_pred, y_test)}")
print(f"Classification Report = {classification_report(y_pred, y_test)}")
print(f"Confusion Matrix = {confusion_matrix(y_pred, y_test)}")
print(f"F1 Score = {f1_score(y_pred, y_test)}")
print(f"Precision Score = {precision_score(y_pred, y_test)}")
print(f"Recall Score = {recall_score(y_pred, y_test)}")
print()

try:
    if y_pred[0] == 1:
        print(f"Have you been Defaulted = Yes")
    elif y_pred[0] == 0:
        print(f"Have you been Defaulted = No")
except Exception:
    print(f"Error = {Exception}")            
```

```python
plt.figure(figsize=(13,2))
plt.scatter (
    x_test["income_k"],
    x_test["credit_score"],
    x_test["debt_k"],
    x_test["credit_utilization"],
    x_test["late_payments"],
    c = y_pred
)
plt.title("Logisic Regression")
plt.xlabel(X.columns)
plt.ylabel("Credit Card Defaulted")
plt.show()
```
