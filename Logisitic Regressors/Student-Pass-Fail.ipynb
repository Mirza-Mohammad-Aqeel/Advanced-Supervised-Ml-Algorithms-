```python
import pandas as pd
from matplotlib import pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, f1_score, precision_score, recall_score

student_pass_fail = pd.read_csv("Student-Pass-Fail.csv")
student_pass_fail = student_pass_fail.dropna()
student_pass_fail['passed'] = student_pass_fail['passed'].map({"Yes":1,"No":0})
```

```python
X = student_pass_fail.drop(columns="passed")
Y = student_pass_fail["passed"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=43)
```

```python
model = LogisticRegression()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict([[3, 61, 4, 78, 5]])
print(f"Accuracy Score = {accuracy_score(y_pred, y_test)}")
print(f"Classification Report = {classification_report(y_pred, y_test)}")
print(f"Confusion Matrix = {confusion_matrix(y_pred, y_test)}")
print(f"F1 Score = {f1_score(y_pred, y_test)}")
print(f"Precision Score = {precision_score(y_pred, y_test)}")
print(f"Recall Score = {recall_score(y_pred, y_test)}")
print()

try:
    if y_pred[0] == 1:
        print(f"Has the Student passed = Yes")
    elif y_pred[0] == 0:
        print(f"Has the Student passed = No")
except Exception:
    print(f"Error = {Exception}")            
```

```python
plt.figure(figsize=(13,2))
plt.scatter (
    x_test["study_hours"],
    x_test["attendance_pct"],
    x_test["assignments_completed"],
    x_test["previous_score"],
    x_test["sleep_hours"],
    c = y_pred
)
plt.title("Logisic Regression")
plt.xlabel(X.columns)
plt.ylabel("Passed")
plt.show()
```
