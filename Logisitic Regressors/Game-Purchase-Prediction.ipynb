```python
import pandas as pd
from matplotlib import pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, f1_score, recall_score, precision_score

game_prucahse = pd.read_csv("Game-Purchase-Prediction.csv")
game_prucahse = game_prucahse.dropna()
game_prucahse['will_buy'] = game_prucahse['will_buy'].map({"Yes":1,"No":0})
```

```python
X = game_prucahse.drop(columns="spam")
Y = game_prucahse["spam"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=43)
```

```python
model = LogisticRegression()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict([[11, 4, 28, 6, 2.5]])
print(f"Accuracy Score = {accuracy_score(y_pred, y_test)}")
print(f"Classification Report = {classification_report(y_pred, y_test)}")
print(f"Confusion Matrix = {confusion_matrix(y_pred, y_test)}")
print(f"F1 Score = {f1_score(y_pred, y_test)}")
print(f"Precision Score = {precision_score(y_pred, y_test)}")
print(f"Recall Score = {recall_score(y_pred, y_test)}")
print()

try:
    if y_pred[0] == 1:
        print(f"Will you buy the Game = Yes")
    elif y_pred[0] == 0:
        print(f"Will you buy the Game = No")
except Exception:
    print(f"Error = {Exception}")            
```

```python
plt.figure(figsize=(13,2))
plt.scatter (
    x_test["age"],
    x_test["gaming_hours_week"],
    x_test["monthly_income"],
    x_test["previous_purchases"],
    x_test["game_rating"],
    c = y_pred
)
plt.title("Logisic Regression")
plt.xlabel(X.columns)
plt.ylabel("Game Purchase Status")
plt.show()
```
