```python
import pandas as pd
from matplotlib import pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, f1_score, precision_score, recall_score

movie_success_df = pd.read_csv("Movie-Success-Prediction.csv")
movie_success_df = movie_success_df.dropna()
movie_success_df['successful'] = movie_success_df['successful'].map({"Yes":1,"No":0})
```

```python
X = movie_success_df.drop(columns="successful")
Y = movie_success_df["successful"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=43)
```

```python
model = LogisticRegression()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict([[4, 2, 7.9, 15, 108, 3]])
print(f"Accuracy Score = {accuracy_score(y_pred, y_test)}")
print(f"Classification Report = {classification_report(y_pred, y_test)}")
print(f"Confusion Matrix = {confusion_matrix(y_pred, y_test)}")
print(f"F1 Score = {f1_score(y_pred, y_test)}")
print(f"Precision Score = {precision_score(y_pred, y_test)}")
print(f"Recall Score = {recall_score(y_pred, y_test)}")
print()

try:
    if y_pred[0] == 1:
        print(f"Was the movie successful = Yes")
    elif y_pred[0] == 0:
        print(f"Was the movie successful = No")
except Exception:
    print(f"Error = {Exception}")            
```

```python
plt.figure(figsize=(13,2))
plt.scatter (
    x_test["budget_m"],
    x_test["marketing-m"],
    x_test["rating"],
    x_test["cast_size"],
    x_test["runtime_min"],
    x_test["opening_weekend_m"],
    c = y_pred
)
plt.title("Logisic Regression")
plt.xlabel(X.columns)
plt.ylabel("Approved")
plt.show()
```
