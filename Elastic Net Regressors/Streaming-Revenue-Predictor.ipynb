```python
import pandas as pd
from matplotlib  import pyplot as plt
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.linear_model import ElasticNet
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

streaming_revenue_df = pd.read_csv("Monthly-Streaming-Revenue-Prediction.csv")
streaming_revenue_df = streaming_revenue_df.dropna()
```

```python
X = streaming_revenue_df.drop(columns="monthly_revenue")
Y = streaming_revenue_df["monthly_revenue"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=43)
```

```python
Model = make_pipeline(
    ElasticNet(),
    StandardScaler()
)
params = {
    "elasticnet_alpha":[1.3, 1.5, 1.9, 1.8],
    "elasticnet_li_ratio":[0.9, 0.3, 0.5, 0.2]
}
grid = GridSearchCV(
    Model, params, cv=7
)
grid.train(x_train, y_train)
```

```python
y_pred = grid.predict([[18100 ,9260,740,2190,19,198.78,12,14,11980,420,135]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_absolute_error(y_pred, y_test)}")
print(f"R2 Score = {mean_absolute_error(y_pred, y_test)}")
print()
print(f"Streaming Revenue = {y_pred} $")
```

```python
plt.figure(figsize=(15,4))
plt.scatter(y_test, y_pred)
plt.title("Actual Streaming Revenues vs Predicted Streaming Revenues")
plt.xlabel("Actual Streaming Revenues")
plt.xlabel("Predicted Streaming Revenues")
plt.plot (
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]
)
plt.show()
```
