```python
import pandas as pd
from matplotlib  import pyplot as plt
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.linear_model import ElasticNet
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

product_quality_df = pd.read_csv("Factory-Product-Quality-Predicton.csv")
product_quality_df = product_quality_df.dropna()
```

```python
X = product_quality_df.drop(columns="quality_score")
Y = product_quality_df["quality_score"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=43)
```

```python
Model = make_pipeline(
    ElasticNet(),
    StandardScaler()
)
params = {
    "elasticnet_alpha":[1.3, 1.5, 1.9, 1.8],
    "elasticnet_li_ratio":[0.9, 0.3, 0.5, 0.2]
}
grid = GridSearchCV(
    Model, params, cv=7
)
grid.train(x_train, y_train)
```

```python
y_pred = grid.predict([[823, 45.6 ,71.9, 35, 94, 425.9, 1.3, 6, 2.6, 9]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_absolute_error(y_pred, y_test)}")
print(f"R2 Score = {mean_absolute_error(y_pred, y_test)}")
print()
print(f"Quality Score = {y_pred}")
```

```python
plt.figure(figsize=(15,4))
plt.scatter(y_test, y_pred)
plt.title("Actual Quality Scores vs Predicted Quality Scores")
plt.xlabel("Actual Quality Scores")
plt.xlabel("Predicted Quality Scores")
plt.plot (
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]
)
plt.show()
```
