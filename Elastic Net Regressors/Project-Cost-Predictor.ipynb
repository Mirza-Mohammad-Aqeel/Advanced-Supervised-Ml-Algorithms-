```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.linear_model import ElasticNet
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

cost_df = pd.read_csv("Construction-Project-Cost-Prediction.csv")
cost_df = cost_df.dropna
```

```python
X = cost_df.drop(columns="final_cost")
Y = cost_df["final_cost"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.3, random_state=69)
```

```python
Model = make_pipeline(
    StandardScaler(),
    ElasticNet()
)
params = {
    "elasticnet_alpha":[1.9,1.4, 1.5,2.4],
    "elasticnet_li_ratio":[0.9,0.4, 0.8, 0.7]
}
grid = GridSearchCV(
    Model, params, cv=9
)
grid.fit(x_train, y_train)
```

```python
y_pred = grid.predict([[855, 20, 145, 460, 90, 210400, 4540, 15200, 38.9, 1250]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print(f"Best Params = {grid.best_params_}")
print()
print(f"Project Cost = {y_pred} $")
```

```python
plt.figure(figsize=(16,7))
plt.scatter(y_test, y_pred)
plt.title("Actual Project Costs vs Predicted Project Costs")
plt.xlabel("Actual Project Costs")
plt.ylabel("Predicted Project Costs")
plt.plot (
    [y_test.min(), y_test.max()],
)
```
