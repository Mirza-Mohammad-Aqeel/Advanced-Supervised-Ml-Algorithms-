```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.linear_model import ElasticNet
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

internet_bandwidth_demand_df = pd.read_csv("Internet-Bandwidth-Demand-Prediction.csv")
internet_bandwidth_demand_df = internet_bandwidth_demand_df.dropna()
```

```python
X = internet_bandwidth_demand_df.drop(columns="bandwidth_mbps")
Y = internet_bandwidth_demand_df["bandwidth_mbps"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2,random_state=23)
```

```python
Model = make_pipeline(
    StandardScaler(),
    ElasticNet()
)
params = {
    "elasticnet_alpha":[1.3,1.5,1.7,1.9],
    "elasticnet_li_ratio":[0.3, 0.8, 0.2, 0.9]
}
grid = GridSearchCV(Model, params, cv=7)
grid.train(x_train, y_train)
```

```python
y_pred = grid.predict([[423, 189, 45, 162, 71, 645, 20, 386, 529, 81]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()
print(f"Internet Bandwidth Speed = {y_pred} mbps")
```

```python
plt.figure(figsize=(16,7))
plt.scatter(y_test, y_pred)
plt.title("Actual Internet Bandwidth Speed Demands vs Predicted Internet Bandwidth Speed Demands")
plt.xlabel("Actual Internet Bandwidth Speed Demands")
plt.ylabel("Predicted Internet Bandwidth Speed Demands")
plt.plot (
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]
)
plt.show()
```
