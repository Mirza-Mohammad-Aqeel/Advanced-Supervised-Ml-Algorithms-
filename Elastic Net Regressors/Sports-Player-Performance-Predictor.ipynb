```python
import pandas as pd
from matplotlib  import pyplot as plt
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.linear_model import ElasticNet
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

player_df = pd.read_csv("Sports-Player-Performance-Prediction.csv")
player_df = player_df.dropna()
```

```python
X = player_df.drop(columns="performance_score")
Y = player_df["performance_score"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=43)
```

```python
Model = make_pipeline(
    ElasticNet(),
    StandardScaler()
)
params = {
    "elasticnet_alpha":[1.3, 1.5, 1.9, 1.8],
    "elasticnet_li_ratio":[0.9, 0.3, 0.5, 0.2]
}
grid = GridSearchCV(
    Model, params, cv=7
)
grid.train(x_train, y_train)
```

```python
y_pred = grid.predict([[23, 145.6 ,71.9, 1835, 94, 25.9, 61.3, 46, 22.6, 9]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_absolute_error(y_pred, y_test)}")
print(f"R2 Score = {mean_absolute_error(y_pred, y_test)}")
print()
print(f"Player Performance Score = {y_pred}")
```

```python
plt.figure(figsize=(15,4))
plt.scatter(y_test, y_pred)
plt.title("Actual Player Performance Scores vs Predicted Player Performance Scores")
plt.xlabel("Actual Player Performance Scores")
plt.xlabel("Predicted Player Performance Scores")
plt.plot (
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]
)
plt.show()
```
