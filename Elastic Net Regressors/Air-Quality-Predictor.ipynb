```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import ElasticNet
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import GridSearchCV

air_quality_df = pd.read_csv("Air-Quality-Prediction.csv")
air_quality_df = air_quality_df.dropna()
```

```python
X = air_quality_df.drop(columns="pm25")
Y = air_quality_df["pm25"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=34)
```

```python
Model = make_pipeline(
    StandardScaler(),
    ElasticNet(alpha=1.4, l1_ratio=0.87)
)

params = {
    "elasticnet_alpha":[0.01, 1.09, 2.01, 0.03],
    "elasticnet_li_ratio":[0.80, 0.91, 0.78, 0.32]
}
grid = GridSearchCV(
    Model,
    params,
    cv=6
)
grid.fit(x_train, y_train)
```

```python
y_pred = grid.predict([[1.4, 13.5, 4, 40, 31.5, 56, 79, 12.6, 14, 1056, 23]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print(f"Best parameters = {grid.best_params_}")
print()
print(f"Air Quality Score = {y_pred}")
```

```python
plt.figure(figsize=(14,5))
plt.scatter(y_test, y_pred)
plt.title("Actual Air Quality Scores vs Predicted Air Quality Scores")
plt.xlabel("Actual Air Quality Scores")
plt.ylabel("Predicted Air Quality Scores")
plt.plot (
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]
)
plt.show()
```
