```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import ElasticNet
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.preprocessing import StandardScaler

city_housing_df = pd.read_csv("City-Housing-Demand-Prediction.csv")
city_housing_df = city_housing_df.dropna()
```

```python
X = city_housing_df.drop(columns="demand_index")
Y = city_housing_df["demand_index"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=34)
```

```python
Model = make_pipeline(
    StandardScaler(),
    ElasticNet()
)
params = {
    "elasticnet_alpha":[1.3, 1.5, 1.1, 2.9],
    "elasticnet_li_ratio":[0.9, 0.3, 0.4, 0.5]
}
grid = GridSearchCV(
    Model,
    params,
    cv=6
)
grid.fit(x_train, y_train)
```

```python
y_pred = grid.predict([[1.4, 4320, 2.6, 76500, 7.8, 2400, 1350, 1200, 3.4, 5.6]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print(f"Best parameters = {grid.best_params_}")
print()
print(f"Demand Index = {y_pred}")
```

```python
plt.figure(figsize=(15,4))
plt.scatter(y_test, y_pred)
plt.title("Actual Demand Indices vs Predicted Indices")
plt.xlabel("Actual Demand Indices")
plt.ylabel("Predicted Indices")
plt.plot (
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
)
plt.show()
```
