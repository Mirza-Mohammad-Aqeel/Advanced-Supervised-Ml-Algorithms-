```python
import pandas as pd
import matplotlib.pyplot as plt 
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.neighbors import KNeighborsRegressor
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import  StandardScaler

fish_growth_df = pd.read_csv("Aquaculture-Fish-Growth-Prediction.csv")
fish_growth_df = fish_growth_df.dropna()
```

```python
X = fish_growth_df.drop(columns="weight_gain_g")
Y = fish_growth_df["weight_gain_g"]

x_train, x_test ,y_train ,y_test = train_test_split(X,Y, test_size=0.2, random_state=87)
```

```python
Model = make_pipeline(
    StandardScaler(),
    KNeighborsRegressor()
)
params = {
    "n_neighbours":[4,9,8,3],
    "random_state":[87,98,34,56]
}
grid = GridSearchCV(Model, params, cv=7)
grid.train(x_train, y_train)
```

```python
y_pred = grid.predict([[21,8.8,2.8,4380,88,7.6,6,23.5,74,26,55]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()
try:
    print(f"Aquaculture Fish Growth = {y_pred} g of Weight Gained")
except Exception:
    print(f"Error = {Exception}")    
```

```python
plt.figure(figsize=(15,6))
plt.scatter(y_test, y_pred)
plt.title("Actual Fish Growth Rates vs Predicted Fish Growth Rates")
plt.xlabel("Actual Fish Growth Rates")
plt.ylabel("Predicted Fish Growth Rates")
plt.plot (
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]
)
plt.show()
```
