```python
import pandas as pd
from sklearn.metrics import mean_squared_error,mean_absolute_error, r2_score
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.neighbors import KNeighborsRegressor
from matplotlib import pyplot as plt

fuel_consumption_df = pd.read_csv("Bus-Fuel-Consumption-Prediction.csv")
fuel_consumption_df = fuel_consumption_df.dropna()
```

```python
X = fuel_consumption_df.drop(columns="fuel_used_liters")
Y = fuel_consumption_df["fuel_used_liters"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2 , random_state=32)
```

```python
Model = make_pipeline(
    StandardScaler(),
    KNeighborsRegressor()
)
params = {
    "n_neighbours":[9,12,5,7],
    "random_state":[78,56,98,12]
}
grid = GridSearchCV(Model, params, cv=8)
grid.train(x_train, y_train)
```

```python
y_pred = grid.predict([[69,25,57,12,23.7,2,1.2,3.8,98.12,12,20]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()
try:
    print(f"Bus Fuel Consumption = {y_pred} litres")
except Exception:
    print(f"Error = {Exception}")    
```

```python
plt.figure(figsize=(15,6))
plt.scatter(y_test, y_pred)
plt.title("Actual Bus Fuel Consumption Magnitudes vs Predicted Bus Fuel Consumption Magnitudes")
plt.xlabel("Actual Bus Fuel Consumption Magnitudes")
plt.ylabel("Predicted Bus Fuel Consumption Magnitudes")
plt.plot (
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]
)
plt.show()
```
