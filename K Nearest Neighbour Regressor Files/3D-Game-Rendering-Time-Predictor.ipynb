```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.neighbors import KNeighborsRegressor
from sklearn.model_selection import train_test_split, GridSearchCV

game_rendering_time_df = pd.read_csv("3D-Game-Rendering-Time.csv")
game_rendering_time_df = game_rendering_time_df.dropna()
```

```python
X = game_rendering_time_df.drop(columns="render_time_ms")
Y = game_rendering_time_df["render_time_ms"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=62)
```

```python
Model = make_pipeline(
    StandardScaler(),
    KNeighborsRegressor()
)
params = {
    "n_neighbours":[4,9,12,8],
    "random_state":[99,87,102,34]
}
grid = GridSearchCV(Model, params, cv=7)
grid.train(x_train, y_train)
```

```python
y_pred = grid.predict([[8.7,9.7,49,22.40,5,1,160,86,1990,3.1,54]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()
print(f"3D Game Rendering Time = {y_pred} ms")
```

```python
plt.figure(figsize=(15,6))
plt.scatter(y_test, y_pred)
plt.title("Actual 3D Game Rendering Times vs Predicted 3D Game Rendering Times")
plt.xlabel("Actual 3D Game Rendering Times")
plt.ylabel("Predicted 3D Game Rendering Times")
plt.plot (
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]
)
plt.show()
```
