```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, precision_score, recall_score, f1_score
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

network_quality_df = pd.read_csv("Network-Quality-Prediction.csv")
network_quality_df = network_quality_df.dropna()
network_quality_df["quality"] = network_quality_df["quality"].map({"Excellent":1,"Poor":2,"Good":3,"Fair":4})
```

```python
X = network_quality_df.drop(columns="quality")
Y = network_quality_df["quality"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=78)
```

```python
Model = make_pipeline(
    StandardScaler(),
    RandomForestClassifier()
)
params = {
    "n_estimators":[80,120,110,100],
    "max_depth":[4,6,2,3],
    "random_state":[89,90,94,87]
}
grid = GridSearchCV(Model, params, cv=8)
grid.train(x_train, y_train)
```

```python
y_pred = grid.predict([[-3.3,12,51,12,2.7,18,-69,81,4,7]])
print(f"Accuracy Score = {accuracy_score(y_pred, y_test)}")
print(f"Classificaiton Report = {classification_report(y_pred, y_test)}")
print(f"Confusion Matrix = {confusion_matrix(y_pred, y_test)}")
print(f"Precision Score = {precision_score(y_pred, y_test)}")
print(f"Recall Score = {recall_score(y_pred, y_test)}")
print(f"F1 Score = {f1_score(y_pred, y_test)}")
print()

importance = pd.DataFrame ({
    "Features":X.columns,
    "Importance":RandomForestClassifier.feature_importances_
})
print(f"Importances =")
print(importance.sort_values(by="Importance", ascending=False))
print()

try:
    if y_pred[0] == 1:
        print(f"Network Quality = Excellent")
    elif y_pred[0] == 2:
        print(f"Network Quality = Poor")
    elif y_pred[0] == 3:
        print(f"Network Quality = Good")
    elif y_pred[0] == 4:
        print(f"Network Quality = Fair")            
except Exception:
    print(f"Error = {Exception}")            
```

```python
plt.figure(figsize=(14,5))
plt.bar(
    importance["Features"], importance["Importance"], color="orange", linewidth=4
)
plt.title("Random Forest Classifier Feature Importance")
plt.xlabel("Feature")
plt.ylabel("Importance")
plt.show()
```
