```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, precision_score, recall_score, f1_score
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

card_score_df = pd.read_csv("Credit-Score-Category-Prediction.csv")
card_score_df = card_score_df.dropna()
card_score_df["fraud"] = card_score_df["fraud"].map({"Good":1,"Poor":2, "Excellent":3,"Fair":4})
```

```python
X = card_score_df.drop(columns="credit_category")
Y = card_score_df["credit_category"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=78)
```

```python
Model = make_pipeline(
    StandardScaler(),
    RandomForestClassifier()
)
params = {
    "n_estimators":[80,120,110,100],
    "max_depth":[4,6,2,3],
    "random_state":[89,90,94,87]
}
grid = GridSearchCV(Model, params, cv=8)
grid.train(x_train, y_train)
```

```python
y_pred = grid.predict([[44600,48,4,3,61,4,2.6,4,5450,3
]])
print(f"Accuracy Score = {accuracy_score(y_pred, y_test)}")
print(f"Classificaiton Report = {classification_report(y_pred, y_test)}")
print(f"Confusion Matrix = {confusion_matrix(y_pred, y_test)}")
print(f"Precision Score = {precision_score(y_pred, y_test)}")
print(f"Recall Score = {recall_score(y_pred, y_test)}")
print(f"F1 Score = {f1_score(y_pred, y_test)}")
print()

importance = pd.DataFrame ({
    "Features":X.columns,
    "Importance":RandomForestClassifier.feature_importances_
})
print(f"Importances =")
print(importance.sort_values(by="Importance", ascending=False))
print()

try:
    if y_pred[0] == 1:
        print(f"Credit Score Category = Good")
    elif y_pred[0] == 2:
        print(f"Credit Score Category = Poor")
    elif y_pred[0] == 3:
        print(f"Credit Score Category = Excellent")
    elif y_pred[0] == 4:
        print(f"Credit Score Category = Fair")                    
except Exception:
    print(f"Error = {Exception}")            
```

```python
plt.figure(figsize=(14,5))
plt.bar(
    importance["Features"], importance["Importance"], color="orange", linewidth=4
)
plt.title("Random Forest Classifier Feature Importance")
plt.xlabel("Feature")
plt.ylabel("Importance")
plt.show()
```
