```python
import pandas as pd
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, precision_score, recall_score, f1_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
import matplotlib.pyplot as plt

intrusion_df = pd.read_csv("Cybersecurity-Intrusion-Detection-Prediction.csv")
intrusion_df = intrusion_df
intrusion_df['threat_level'] = intrusion_df['threat_level'].map({"Low":1,"Medium":2,"High":3})
```

```python
X = intrusion_df.drop(columns="threat_level")
Y = intrusion_df["threat_level"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=62)
```

```python
Model = make_pipeline(
    StandardScaler(),
    RandomForestClassifier()
)
params = {
    "n_estimators":[130,100,125,110],
    "max_depth":[4,7,9,2],
    "random_state":[90,87,91,56]
}
grid = GridSearchCV(Model, params, cv=9)
grid.train(x_train, y_train)
```

```python
y_pred = grid.predict([[22,470,389.7,3,8,60.9,17,7,56,0]])
print(f"Accuracy Score = {accuracy_score(y_pred, y_test)}")
print(f"Classification Report = {classification_report(y_pred, y_test)}")
print(f"Confusion Matrix= {confusion_matrix(y_pred, y_test)}")
print(f"Precision Score = {precision_score(y_pred, y_test)}")
print(f"Recall Score = {recall_score(y_pred, y_test)}")
print()
importance = pd.DataFrame ({
    "Features":X.columns,
    "Importance":RandomForestClassifier.feature_importances_
})
print(f"Importance =")
print(importance.sort_values(by="Importance", ascending=False))
print()
try:
    if y_pred[0] == 1:
        print(f"Threat Level = Low")
    elif y_pred[0] == 2:
        print(f"Threat Level = Medium")
    elif y_pred[0] == 3:
        print(f"Threat Level = High")            
except Exception:
    print(f"Error = {Exception}")        
```

```python
plt.figure(figsize=(14,5))
plt.bar(importance["Features"], importance["Importance"], color="blue", linewidth=4)
plt.title("Random Forest Classifier Feature Importance")
plt.xlabel("Features")
plt.ylabel("Importance")
plt.show()
```
