Importing & Cleaning the Data

```python
import pandas as pd
from matplotlib import pyplot as plt

from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor, plot_tree

employee_salary_df = pd.read_csv("Employee-Salary-Prediction.csv")
employee_salary_df = employee_salary_df.dropna()
```

Splitting the Data

```python
X = employee_salary_df.drop(columns="annual_salary")
Y = employee_salary_df["annual_salary"]

x_train, x_test, y_train, y_test = train_test_split(X,Y,test_size=0.2, random_state=35)
```

Creating & Training the Model

```python
model = DecisionTreeRegressor(
    criterion="squared_error",
    max_depth=5
    random_state=91
)
model.fit(x_train, y_train)
```

Making Prediction & Evaluating Accuracy

```python
y_pred = model.predict([[29,8,2,4,29]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()

importance = pd.DataFrame({
    "Feature" : X.columns,
    "Importance":model.feature_importances_
})

print(f"Importance =")
print(importance.sort_values(by="Importance", ascending=False))
print()

print(f"Employee Salary = {y_pred}")
```

Visualizing the Model

```python
plt.figure(figsize=(11,3))
plot_tree(model, feature_names=X.columns, filled=True, rounded=True)
plt.title("Decision Tree Regressor")
plt.show()
```
