Importing & Cleaning Data

```python
import pandas as pd 
import matplotlib.pyplot as plt

from sklearn.tree import DecisionTreeRegressor, plot_tree
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

apartment_rent_dataframe = pd.read_csv("Apartment-Rent-Prediction.csv")
apartment_rent_dataframe = apartment_rent_dataframe.dropna()
```

Splitting the Data

```python
X = apartment_rent_dataframe.drop(columns="monthly_rent")
Y = apartment_rent_dataframe["monthly_rent"]

x_train, x_test, y_train, y_test = train_test_split(X,Y,test_size=0.2, random_state=42)
```

Creating & Training the Model

```python
model = DecisionTreeRegressor(criterion="squared_error", max_depth=4, random_state=42)
model.fit(x_train, y_train)
```

Making Predictions & Evaluating Accuracy Score

```python
y_pred = model.predict([[180,9,4,2,6]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
importance = pd.DataFrame({
    "Feature" : X.columns,
    "Importance" : model.feature_importances_
})
print(f"Feature Importances")
print(importance.sort_values(by="Importance", ascending=False))
print()
print(f"Apartment Rent = {y_pred}")
```

Visualizing the Model

```python
plt.figure(figsize=(18,9))
plot_tree(model, feature_names= X.columns, filled=True, rounded=True)
plt.title("Decision Tree Regressor")
plt.show()
```
