Importing & Cleaning the Data

```python
import pandas as pd 
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.tree import DecisionTreeRegressor, plot_tree

electricity_consumption_df = pd.read_csv("Electricity-Consumption-Prediction.csv")
electricity_consumption_df = electricity_consumption_df.dropna()
```

Splitting the Data

```python
X = electricity_consumption_df.drop(columns="monthly_kwh")
Y = electricity_consumption_df["monthly_kwh"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=12)
```

Creating & Training the Model

```python
model = DecisionTreeRegressor (
    criterion="squared_error",
    max_depth=6,
    random_state=24
)
model.fit(x_train, y_train)
```

Making Prediction & Evaluating Accuracy

```python
y_pred = model.predict([[4,9,1,670,3,1]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()
importance = pd.DataFrame({
    "Feature":X.columns,
    "Importance":model.feature_importances_
})
print(f"Importance =")
print(importance.sort_values(by="Importance", ascending=False))
print()

print(f"Electricity Consumption = {y_pred}")
```

Visualizing the Model

```python
plt.figure(figsize=(15,4))
plot_tree(model, feature_names=X.columns, filled=True, rounded=True)
plt.title("Decision Tree Regressor")
plt.show()
```
