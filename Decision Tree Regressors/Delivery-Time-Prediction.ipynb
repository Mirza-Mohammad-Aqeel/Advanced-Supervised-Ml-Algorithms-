Importing & Cleaning Data

```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.tree import DecisionTreeRegressor, plot_tree
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.model_selection import train_test_split

delivery_time_df = pd.read_csv("Delivery-Time-Prediction.csv")
delivery_time_df = delivery_time_df.dropna()
```

Splitting the Data

```python
X = delivery_time_df.drop(columns="delivery_time_min")
Y = delivery_time_df["delivery_time_min"]

x_train, x_test, y_train, y_test = train_test_split(X,Y,test_size=0.2, random_state=67)
```

Creating & Training the Model

```python
model = DecisionTreeRegressor(criterion="squared_error", max_depth=5, random_state=80)
model.fit(x_train, y_train)
```

Making Predictions & Evaluating Accuracy

```python
y_pred = model.predict([[5.4,6,7.8,3,1]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
importance = pd.DataFrame ({
    "Feature" : X.columns,
    "Importance" : model.feature_importances_
})
print("Importance = ")
print(importance.sort_values(by="Importance", ascending=False))
print()

print(f"Delivery Time = {y_pred}")
```

Visualizing the Model

```python
plt.figure(figsize=(12,5))
plot_tree(model, feature_names=X.columns, filled=True, rounded=True)
plt.title("Decision Tree Regressor")
plt.show()
```
