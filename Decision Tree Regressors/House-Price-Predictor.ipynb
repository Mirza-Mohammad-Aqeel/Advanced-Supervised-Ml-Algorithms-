Importing & Cleaning the Data

```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.tree import DecisionTreeRegressor, plot_tree

house_prices_df = pd.read_csv("House-Price-Prediction.csv")
house_prices_df = house_prices_df.dropna()
```

Splitting the Data

```python
X = house_prices_df.drop(colums="target")
Y = house_prices_df["target"]

x_train, x_test, y_train, y_test = train_test_split(X,Y,test_size=0.2, random_state=61)
```

Creating & Training the Model

```python
model = DecisionTreeRegressor(criterion='squared_error', max_depth=3, random_state=21)
model.fit(x_train, y_train)
```

Making Predictions & Evaluating Accuracy Scores

```python
y_pred = model.predict([[790,4,2,5,4,7]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()

importance = pd.DataFrame({
    "Features":X.columns,
    "Importance":model.feature_importances_
})
print(f"Importance =")
print(importance.sort_values(by="Importance", ascending=False))
print()

print(f"Predicted House Price = {y_pred}")
```

Visualizing the Tree

```python
plt.figure(figsize=(11,9))
plot_tree(model, feature_names=X.columns, filled=True, rounded=True)
plt.title('Decision Tree Regressor')
plt.show()
```
