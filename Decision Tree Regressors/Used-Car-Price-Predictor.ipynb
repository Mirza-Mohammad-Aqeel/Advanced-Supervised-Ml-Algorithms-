Importing & Cleaning the Data

```python
import pandas as pd 
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor, plot_tree
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

car_prices_df = pd.read_csv("Used-Car-Price-Prediction.csv")
car_prices_df = car_prices_df.dropna()
```

Splitting the Data

```python
X = car_prices_df.drop(columns="price_usd")
Y = car_prices_df["price_usd"]

x_train, x_test , y_train, y_test = train_test_split(X,Y,test_size=0.2, random_state=23)
```

Creating & Training the Model

```python
model = DecisionTreeRegressor(criterion="squared_error", max_depth=3, random_state=46)
model.fit(x_train, y_train)
```

Making Predictions & Evaluating Accuracy

```python
y_pred = model.predict([[2019,198100, 1300, 450,2,5]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()

importance = pd.DataFrame ({
    "Feature":X.columns,
    "Importance":model.feature_importances_
})
print(f"Importance =")
print(importance.sort_values(by="Importance", ascending=False))
print()

print(f"Car Price = {y_pred}")
```

Visualizing the Tree

```python
plt.figure(figsize=(19,8))
plot_tree(model, feature_names=X.columns, rounded=True, filled=True)
plt.title("Decision Tree Regressor")
plt.show()
```
