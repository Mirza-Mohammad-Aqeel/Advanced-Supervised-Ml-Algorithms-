Importing & Cleaning the Data

```python
import pandas as pd 
from matplotlib import pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.tree import DecisionTreeRegressor, plot_tree

student_score_df = pd.read_csv("Student-Score-Prediction.csv")
student_score_df = student_score_df.dropna()
```

Splitting the Data

```python
X = student_score_df.drop(columns="final_score")
Y = student_score_df["final_score"]

x_train, x_test, y_train, y_test = train_test_split(X,Y,test_size=0.2, random_state=89)
```

Creating & Training the Model 

```python
model = DecisionTreeRegressor(criterion="squared_error",max_depth=3, random_state=12)
model.fit(x_train, y_train)
```

Making Predictions & Evaluating Accuracy

```python
y_pred = model.predict([[3,45,89,9,8]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()

importance = pd.DataFrame ({
    "Feature":X.columns,
    "Importance":model.feature_importances_
})
print(f"Importance =")
print(importance.sort_values(by="Importance", ascending=False))
print()

print(f"Score = {y_pred}")
```
