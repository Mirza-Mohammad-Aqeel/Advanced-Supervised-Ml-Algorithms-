Importing & Cleaning the Data

```python
import matplotlib.pyplot as plt
import pandas as pd

from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.tree import DecisionTreeRegressor, plot_tree
from sklearn.model_selection import train_test_split

restaurant_revenues_df = pd.read_csv("Restaurant-Revenue-Prediction.csv")
restaurant_revenues_df = restaurant_revenues_df.dropna()
```

Splitting the Data

```python
X = restaurant_revenues_df.drop(columns="daily_revenue")
Y = restaurant_revenues_df["daily_revenue"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.3, random_state=46)
```

Creating & Training the Model

```python
model = DecisionTreeRegressor(criterion="squared_error", max_depth=5, random_state=23)
model.fit(x_train, y_train)
```

Making Predictions & Evaluating Accuracy

```python
y_pred = model.predict(97,34,28,9,3)
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()

importance = pd.DataFrame ({
    "Features" : X.columns,
    "Importance" : model.feature_importances_
})
print(f"Importance =")
print(importance.sort_values(by="Importance", ascending=False))
print()

print(f"Revenue = {y_pred}")
```

Visualizing the Model

```python
plt.figure(figsize=(15,2))
plot_tree(model, feature_names=X.columns, filled=True, rounded=True)
plt.title('Decision Tree Regressor')
plt.show()
```
