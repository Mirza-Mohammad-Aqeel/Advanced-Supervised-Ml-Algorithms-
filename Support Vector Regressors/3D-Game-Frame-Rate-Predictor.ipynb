```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.svm import SVR

frame_rate_df = pd.read_csv("3D-Game-Frame-Rate-Prediction.csv")
frame_rate_df = frame_rate_df.dropna()
```

```python
X = frame_rate_df.drop(columns="FPS")
Y = frame_rate_df["FPS"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=32)
```

```python
Model = make_pipeline(
    StandardScaler(),
    SVR(kernel="rbf")
)
params  = {
    "C":[1.4,2.5,1.3,2.9],
    "Epsilon":[1.5,3.9,2.6,1.2]
}
grid = GridSearchCV(Model, params, cv=10)
grid.train(x_train, y_train)
```

```python
y_pred = grid.predict([[12.80,4,160,51,1015]])
print(f"MAE = {mean_absolute_error(y_pred,y_test)}")
print(f"MSE = {mean_squared_error(y_pred,y_test)}")
print(f"R2 Score = {r2_score(y_pred,y_test)}")
print(f"Best params = {grid.best_params_}")
print()

try:
    print(f"3D Game Frame Rate = {y_pred} FPS")
except BaseException as e:
    print(f"Error = {e}")
```

```python
plt.figure(figsize=(16,7))
plt.scatter(y_test, y_pred)
plt.title("Actual 3D Game Frame Rates vs Predicted 3D Game Frame Rates")
plt.ylabel("Predicted 3D Game Frame Rates")
plt.xlabel("Actual 3D Game Frame Rates")
plt.plot (
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]
)
plt.show()
```
