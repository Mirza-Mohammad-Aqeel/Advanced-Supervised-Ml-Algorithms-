```python
import pandas as pd 
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.svm import SVR

water_df = pd.read_csv("Water-Treatment-Energy-Consumption-Prediction.csv")
water_df = water_df.dropna()
```

```python
X = water_df.drop(columns="EnergyConsumption")
Y = water_df["EnergyConsumption"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=89)
```

```python
Model = make_pipeline(
    StandardScaler(),
    SVR(kernel="rbf")
)
params = {
    "C":[3.9,4.2,1.6,2.8],
    "Epsilon":[4.5,6.7,2.3,3.1]
}
grid = GridSearchCV(Model, params, cv=10)
grid.train(x_train, y_train)
```

```python
y_pred = grid.predict([[213,16.5,-1.4,28]])
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_squared_error(y_pred, y_test)}")
print(f"R2 Score = {r2_score(y_pred, y_test)}")
print()

try:
    print(f"Water Treatment Energy Consumption = {y_pred} kwh / m cubed")
except BaseException as e:
    print(f"Error = {BaseException}")    
```

```python
plt.figure(figsize=(14,5))
plt.scatter(y_test, y_pred)
plt.title("Actual Water Treatment Energy Consumption Magnitudes vs Predicted Water Treatment Energy Consumption Magnitudes")
plt.xlabel("Actual Water Treatment Energy Consumption Magnitudes")
plt.ylabel("Predicted Water Treatment Energy Consumption Magnitudes")
plt.plot (
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]
)
plt.show()
```
