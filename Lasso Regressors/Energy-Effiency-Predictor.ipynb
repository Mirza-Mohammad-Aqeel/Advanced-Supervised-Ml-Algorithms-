```python
import pandas as pd
import matplotlib.pyplot as plt 
from sklearn.linear_model import Lasso
from sklearn.metrics import mean_squared_error,mean_absolute_error, r2_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

energy_effiency = pd.read_csv("Building-Energy-Effiency-Prediction.csv")
energy_effiency = energy_effiency.dropna()
```

```python
X = energy_effiency.drop(columns="electricity_kwh")
Y = energy_effiency["electricity_kwh"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=32)
```

```python
model = make_pipeline(
    StandardScaler(),
    Lasso(alpha=2.3)
)
model.fit(x_train, y_train)
```

```python
y_pred = model.predict([[112.4, 435, 64.7, 32.9,  69, 114, 13, 178, 46, 120]])
print(f"Model Coeffient = {model.coef_}")
print(f"Model Intercept = {model.intercept_}")
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_absolute_error(y_pred, y_test)}")
print(f"R2 Score = {mean_absolute_error(y_pred, y_test)}")
print()
print(f"Electricity Voltage = {y_pred} kwh")
```

```python
plt.figure(figsize=(15))
plt.scatter(y_test, y_pred)
plt.title("Actual Electricity Voltages vs Predicted Electricity Voltages")
plt.xlabel("Actual Electricity Voltages")
plt.ylabel("Predicted Electricity Voltages")
plt.plot (
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
)
plt.show()
```
