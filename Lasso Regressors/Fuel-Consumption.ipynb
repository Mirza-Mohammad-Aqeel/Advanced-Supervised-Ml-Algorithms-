```python
import pandas as pd
import matplotlib.pyplot as plt 
from sklearn.linear_model import Lasso
from sklearn.metrics import mean_squared_error,mean_absolute_error, r2_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

fuel_df = pd.read_csv("Flight-Fuel-Consumption.csv")
fuel_df = fuel_df.dropna()
```

```python
X = fuel_df.drop(columns="fuel_used_liters")
Y = fuel_df["fuel_used_liters"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=32)
```

```python
model = make_pipeline(
    StandardScaler(),
    Lasso(alpha=2.3)
)
model.fit(x_train, y_train)
```

```python
y_pred = model.predict([[2.4, 435, 64.7, 32.9,  69, 14, 13, 8, 46, 120]])
print(f"Model Coeffient = {model.coef_}")
print(f"Model Intercept = {model.intercept_}")
print(f"MAE = {mean_absolute_error(y_pred, y_test)}")
print(f"MSE = {mean_absolute_error(y_pred, y_test)}")
print(f"R2 Score = {mean_absolute_error(y_pred, y_test)}")
print()
print(f"Fuel Used = {y_pred} litres")
```

```python
plt.figure(figsize=(15))
plt.scatter(y_test, y_pred)
plt.title("Actual Fuel Litres vs Predicted Fuel Litres")
plt.xlabel("Actual Fuel Litres")
plt.ylabel("Predicted Fuel Litres")
plt.plot (
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
)
plt.show()
```
