Importing & Cleaning the Model

```python
import pandas as pd
from matplotlib import pyplot as plt

from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree

health_risk_df = pd.read_csv("Health-Risk.csv")
health_risk_df = health_risk_df.dropna()

health_risk_df['class'] = health_risk_df['class'].map ({"Low":1,"Medium":2,"High":3})

risks = {
    1:"Low",
    2:"Medium",
    3:"High"
}
```

Splitting the Data

```python
X = health_risk_df.drop(columns="class")
Y = health_risk_df["class"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=59)
```

Creating & Training the Model

```python
model = DecisionTreeClassifier(criterion="gini", max_depth=5, random_state=8)
model.fit(x_train, y_train)
```

Making Predictions & Evaluating Accuracy

```python
y_pred = model.predict([[24,19.5,6.9, 67, 9.1, 3]])
print(f"Accuracy Score = {accuracy_score(y_pred, y_test)}")
print(f"Classification Report = {classification_report(y_pred, y_test)}")
print(f"Confusion Matrix = {confusion_matrix(y_pred, y_test)}")
print()

importance = pd.DataFrame ({
    "Feature":X.columns,
    "Importance":model.feature_importances_
})
print(f"Importance =")
print(importance.sort_values(by="Importance", ascending=False))
print()

try:
    if y_pred[0] == 1:
        print(f"Health risk = {risks[1]}")
    elif y_pred[0] == 2:
        print(f"Health risk = {risks[2]}")
    elif y_pred[0] == 3:
        print(f"Health risk = {risks[3]}")     
except Exception:
    print(F"Error = {Exception}")               
```

Visualizing the Model

```python
plt.figure(figsize=(12,3))
plot_tree(model, feature_names=X.columns, class_names=model.classes_astype(str), rounded=True, filled=True)
plt.title("Decision Tree Classifier")
plt.show()
```
