Importing & Cleaning the Data

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.model_selection import train_test_split
import pandas as pd
import matplotlib.pyplot as plt

game_success_df = pd.read_csv("Game-Success.csv")
game_success_df = game_success_df.dropna()

game_success_df['success'] = game_success_df['success'].map({"Yes":1,"No":0})
```

Splitting the Data

```python
X = game_success_df.drop(columns="success")
Y = game_success_df["success"]

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=45)
```

Creating & Training the Model

```python
model = DecisionTreeClassifier(criterion="gini", max_depth=5, random_state=20)
model.fit(x_train, y_train)
```

Making Predictions & Evaluating Prediction

```python
y_pred = model.predict([[150,23,6.9,7,98]])
print(f"Accuracy Score = {accuracy_score(y_pred, y_test)}")
print(f"Classification Report = {classification_report(y_pred, y_test)}")
print(f"Confusion Matrix = {confusion_matrix(y_pred, y_test)}")
print()

importance = pd.DataFrame ({
    "Feature":X.columns,
    "Importance":model.feature_importances_
})
print(f"Importance")
print(importance.sort_values(by="Importance",ascending=False))
print()

try:
    if y_pred[0] == 1:
        print(f"Will the game succeed = Yes")
    elif y_pred[0] == 0:
            print(f"Will the game succeed = No")
except Exception:
    print(f"Error = {Exception}")             
```

Visualizing the Model

```python
plt.figure(figsize=(15,4))
plot_tree(model, feature_names=X.columns, class_names=model.classes_astype(str), filled=True, rounded=True)
plt.title("Decision Tree Classifier")
plt.show()
```
